### Introduction

Building a machine learning model involves data collection and preparation, selecting an (appropriate) algorithm, training the model on data, and evaluating its performance through testing and iteration

#### Training data
A model is only as good as the data used to train it. So, data preparation and processing are the most critical stage in making the whole model work as intended. There are several different types of data and they can be generally classified like this.

##### 1. Labelled data:
A dataset(includes both training and test data) in which each instance or example is accompanied by a label or target variable that represents the desired output. These labels are typically provided by human experts or obtained via reliable process

##### 2. Unlabelled data:
Opposite of labelled data, as in a dataset which the instances or examples do not have any associated labels or target variables. The data consists only of input features.

##### 3. Structured data:
Data that is organized and formatted in a predefined manner(Tabular and Time-series data). These type of data is suitable for traditional machine learning algorithms that require well-defined features and labels.

##### 4. Unstructured data:
Data that lacks predefined structure or format(Text, images, audio and video data). This type of data requires more advanced machine learning techniques to extract meaningful patterns and insights.

#### Machine learning Processes
##### 1. Supervised learning
The algorithms are trained on labeled data. The goal is to learn a mapping function that can predict the output fpr a new and unseen input.
##### 2. UnSupervised learning
Algorithms that learn from unlabeled data. Yhe goal is to discover inherent patterns, structures or relationships withing the input data
##### 3. reinforcement learning
In this, the machine is given only a performance score as guidance and semi-supervised learning, where only  a portion of training data is labeled. And then feedback is provided in the form of rewards or penalties for its actions, and the machine learns from this feedback to improve its decision making over time.



#### Inferencing
After the model has been trained, it is time to begin the process of using the information that a model has learned to make predictions or decisions. This is called inferencing. There are two types of inferencing and our use case will determine which inferencing type to use.

##### 1. Batch inferencing
Batch inferencing is when the computer takes a large amount of data, such as images or text, and analyzes it all at once to provide a set of results. This type of inferencing is often used for tasks like data analysis, where the speed of the decision-making process is not as crucial as the accuracy of the results.
##### 2. Real-time inferencing
Real-time inferencing is when the computer has to make decisions quickly, in response to new information as it comes in. This is important for applications where immediate decision-making is critical, such as in chatbots or self-driving cars. The computer has to process the incoming data and make a decision almost instantaneously, without taking the time to analyze a large dataset.
